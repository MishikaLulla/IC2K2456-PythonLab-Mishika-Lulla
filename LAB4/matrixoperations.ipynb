```python
# question 2 — matrix operations
matrix = []
print("enter 9 elements")
for i in range(3):
    row = []
    for j in range(3):
        value = int(input(f"enter element [{i}][{j}]: "))
        row.append(value)
    matrix.append(row)

print("\nmatrix:")

for i in range(3):
    for j in range(3):
        print(matrix[i][j], end=" ")
    print()

total_sum = 0

for i in range(3):
    for j in range(3):
        total_sum += matrix[i][j]

diagonal_sum = 0

for i in range(3):
    diagonal_sum += matrix[i][i]

largest = matrix[0][0]
smallest = matrix[0][0]

for i in range(3):
    for j in range(3):
        if matrix[i][j] > largest:
            largest = matrix[i][j]

        if matrix[i][j] < smallest:
            smallest = matrix[i][j]

print("\nsum of all elements", total_sum)
print("sum of main diagonal", diagonal_sum)
print("largest element", largest)
print("smallest element", smallest)

print("\ntranspose")

for i in range(3):
    for j in range(3):
        print(matrix[j][i], end=" ")
    print()
```
