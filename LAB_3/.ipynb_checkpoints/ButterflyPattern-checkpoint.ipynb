```python
#Write a Python program that takes the number of rows n from the user and prints a butterfly star pattern
n = int(input("enter number of rows"))
for i in range(1, n + 1):
    stars = i
    spaces = 2 * (n - i)

    for j in range(stars):
        print("*", end="")

    for j in range(spaces):
        print(" ", end="")

    for j in range(stars):
        print("*", end="")

    print()

for i in range(n, 0, -1):
    stars = i
    spaces = 2 * (n - i)

    for j in range(stars):
        print("*", end="")

    for j in range(spaces):
        print(" ", end="")

    for j in range(stars):
        print("*", end="")

    print()
```
