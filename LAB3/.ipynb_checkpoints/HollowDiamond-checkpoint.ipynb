```python
#Write a Python program that takes an odd number n from the user and prints a hollow diamond pattern
n = int(input("enter an odd number "))
for i in range(n):
    if i <= n // 2:
        spaces = n // 2 - i
        stars = 2 * i + 1
    else:
        spaces = i - n // 2
        stars = 2 * (n - i) - 1

    for j in range(spaces):
        print(" ", end="")

    for j in range(stars):
        if j == 0 or j == stars - 1 or stars == 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```
