```python
#question 3-program lets the computer guess a number selected by the user.

def reverse_guessing_game():
    while True:
        try:
            lower = int(input("enter the lower limit: "))
            upper = int(input("enter the upper limit: "))

            if lower >= upper:
                print("upper limit must be greater than lower limit")
            else:
                break

        except ValueError:
            print("invalid input")

    print("\nthink of a number between", lower, "and", upper)

    guesses = 0

    while lower <= upper:
        guess = (lower + upper) // 2
        guesses += 1

        print("\ncomputer guess:", guess)
        feedback = input(
            "enter too high too low or correct: "
        ).lower()

        if feedback == "correct":
            print("the computer found your number")
            print("number of guesses:", guesses)
            return

        elif feedback == "too high":
            upper = guess - 1

        elif feedback == "too low":
            lower = guess + 1

        else:
            print("invalid feedback")
            print("please enter too high too low or correct")
            guesses -= 1

    print("the feedback was inconsistent")


reverse_guessing_game()
```
