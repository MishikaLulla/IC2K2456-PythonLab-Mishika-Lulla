```python
#question 5
import random

def atm():
    balance = 5000
    pin = "1234"

    entered_pin = input("enter pin: ")

    if entered_pin != pin:
        print("incorrect pin")
        return

    while True:
        print("\n--- atm ---")
        print("1. check balance")
        print("2. deposit")
        print("3. withdraw")
        print("4. change pin")
        print("5. exit")

        choice = input("enter choice: ")

        if choice == "1":
            print("balance:", balance)

        elif choice == "2":
            try:
                amount = float(input("enter deposit amount: "))

                if amount <= 0:
                    print("amount must be greater than 0")
                else:
                    balance += amount
                    print("deposit successful")
                    print("balance:", balance)

            except ValueError:
                print("invalid amount")

        elif choice == "3":
            try:
                amount = float(input("enter withdrawal amount: "))

                if amount <= 0:
                    print("amount must be greater than 0")
                elif amount > balance:
                    print("insufficient balance")
                else:
                    balance -= amount
                    print("withdrawal successful")
                    print("balance:", balance)

            except ValueError:
                print("invalid amount")

        elif choice == "4":
            old_pin = input("enter current pin: ")

            if old_pin == pin:
                pin = input("enter new pin: ")
                print("pin changed successfully")
            else:
                print("incorrect pin")

        elif choice == "5":
            break

        else:
            print("invalid choice")


def calculate_grade(average):
    if average >= 90:
        return "a"
    elif average >= 75:
        return "b"
    elif average >= 60:
        return "c"
    elif average >= 40:
        return "d"
    else:
        return "f"


def grade_calculator():
    last_marks = None
    last_average = 0
    last_grade = ""

    while True:
        print("\n--- grade calculator ---")
        print("1. enter marks for a new student")
        print("2. view grade of last entered student")
        print("3. exit")

        choice = input("enter choice: ")

        if choice == "1":
            marks = []

            for i in range(5):
                while True:
                    try:
                        mark = float(input(f"enter marks for subject {i + 1}: "))

                        if mark < 0 or mark > 100:
                            print("marks should be between 0 and 100")
                        else:
                            marks.append(mark)
                            break

                    except ValueError:
                        print("invalid input")

            total = 0

            for mark in marks:
                total += mark

            last_marks = marks
            last_average = total / 5
            last_grade = calculate_grade(last_average)

            print("average:", last_average)
            print("grade:", last_grade)

        elif choice == "2":
            if last_marks is None:
                print("no student data available")
            else:
                print("marks:", last_marks)
                print("average:", last_average)
                print("grade:", last_grade)

        elif choice == "3":
            break

        else:
            print("invalid choice")


def guessing_game():
    secret_number = random.randint(1, 100)
    score = 100
    max_attempts = 10

    print("\n--- guessing game ---")
    print("guess the number between 1 and 100")

    for attempt in range(1, max_attempts + 1):
        try:
            guess = int(input("enter your guess: "))

        except ValueError:
            print("invalid input")
            continue

        if guess == secret_number:
            print("correct")
            print("your score is:", score)
            return

        score -= 10

        if guess < secret_number:
            print("your guess is too low")
        else:
            print("your guess is too high")

        if secret_number % 2 == 0:
            print("hint: the number is even")
        else:
            print("hint: the number is odd")

        if secret_number % 5 == 0:
            print("hint: the number is a multiple of 5")
        else:
            print("hint: the number is not a multiple of 5")

    print("you lost")
    print("the secret number was:", secret_number)
    print("your score is: 0")


def main():
    while True:
        print("\n=== main menu ===")
        print("1. atm")
        print("2. grade calculator")
        print("3. guessing game")
        print("4. exit")

        choice = input("enter choice: ")

        if choice == "1":
            atm()

        elif choice == "2":
            grade_calculator()

        elif choice == "3":
            guessing_game()

        elif choice == "4":
            print("program ended")
            break

        else:
            print("invalid choice")


main()
```
