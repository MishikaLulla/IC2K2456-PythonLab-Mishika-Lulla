time complexity: O(mn)
space complexity: O(k)

justify: every element of the m by n matrix is checked once and only the k non zero elements are stored in the triples list

```python
def to_sparse(matrix):
    triples = []

    for r in range(len(matrix)):
        for c in range(len(matrix[0])):
            if matrix[r][c] != 0:
                triples.append((r, c, matrix[r][c]))

    return triples


# Matrix
matrix = [
    [0, 0, 5],
    [0, 8, 0],
    [3, 0, 0]
]

# Call function
result = to_sparse(matrix)

# Print result
print(result)
```
