```python
def fast_power(base, exp):
    if exp == 0:
        return 1

    half = fast_power(base, exp // 2)

    if exp % 2 == 0:
        return half * half

    return half * half * base


result = fast_power(2, 10)

print(result)
```

time complexity: O(log exp)
space complexity: O(log exp)

justify: the exponent is divided by 2 in every recursive call which gives logarithmic calls and logarithmic recursion stack space
