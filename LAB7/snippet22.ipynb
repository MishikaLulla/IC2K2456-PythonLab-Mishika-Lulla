```python
def build_frequency_map(arr):
    freq = {}

    for val in arr:
        freq[val] = freq.get(val, 0) + 1

    return freq


arr = [1, 2, 2, 3, 3, 3, 4]

result = build_frequency_map(arr)

print(result)
```

time complexity: O(n)
space complexity: O(n)

justify: every array element is processed once and the frequency dictionary can contain up to n different values
