```python
def merge_sorted(a, b):
    result = []
    i = j = 0

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    result.extend(a[i:])
    result.extend(b[j:])

    return result


a = [1, 3, 5]
b = [2, 4, 6]

result = merge_sorted(a, b)

print(result)
```

time complexity: O(n + m)
space complexity: O(n + m)

justify: each element from both sorted arrays is processed once and the result list stores all elements
