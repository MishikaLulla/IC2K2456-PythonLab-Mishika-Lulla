```python
def matrix_multiply(a, b):
    n = len(a)
    result = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            for k in range(n):
                result[i][j] += a[i][k] * b[k][j]

    return result



a = [
    [1, 2],
    [3, 4]
]

b = [
    [5, 6],
    [7, 8]
]

# Call 
result = matrix_multiply(a, b)


print(result)
```

```python
time complexity: O(n³) space complexity: O(n²)

justify: three nested loops each run n times while the result matrix requires n² space
```
