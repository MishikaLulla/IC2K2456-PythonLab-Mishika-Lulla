```python
def sum_digits(n):
    if n == 0:
        return 0

    return n % 10 + sum_digits(n // 10)


number = int(input("enter a number: "))

result = sum_digits(number)

print("sum of digits:", result)
```

# snippet 3

time complexity: O(log n)
space complexity: O(log n)

justify: the number is divided by 10 in every recursive call so there are logarithmic calls and the recursive stack also grows logarithmically
