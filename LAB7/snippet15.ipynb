```python
def print_all_subsets(arr):
    n = len(arr)

    for i in range(2 ** n):
        subset = []

        for j in range(n):
            if i & (1 << j):
                subset.append(arr[j])

        print(subset)


arr = [1, 2, 3]

print_all_subsets(arr)
```

time complexity: O(n × 2ⁿ)
space complexity: O(n)

justify: there are 2ⁿ possible subsets and each subset checks all n elements while the temporary subset can contain n elements
