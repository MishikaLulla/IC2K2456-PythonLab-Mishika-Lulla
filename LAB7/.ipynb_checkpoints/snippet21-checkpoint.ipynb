```python
def has_common_element(a, b):
    for x in a:
        for y in b:
            if x == y:
                return True

    return False


a = [1, 2, 3, 4]
b = [5, 6, 3, 8]

result = has_common_element(a, b)

print(result)
```

time complexity: O(n × m)
space complexity: O(1)

justify: in the worst case every element of a is compared with every element of b and no extra data structure is used
