```python
def flatten(matrix):
    flat = []

    for row in matrix:
        for val in row:
            flat.append(val)

    return flat


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

result = flatten(matrix)

print(result)
```

time complexity: O(mn)
space complexity: O(mn)

justify: every value in the m by n matrix is visited once and all values are stored in the flat list
