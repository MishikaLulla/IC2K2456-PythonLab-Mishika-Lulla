```python
def print_pairs(arr):
    n = len(arr)
    result = []

    for i in range(n):
        for j in range(n):
            result.append((arr[i], arr[j]))

    return result


numbers = [1, 2, 3]

result = print_pairs(numbers)

print(result)
```

# snippet 4

time complexity: O(n²)
space complexity: O(n²)

justify: both loops run n times creating n² pairs and the result list stores all these pairs
