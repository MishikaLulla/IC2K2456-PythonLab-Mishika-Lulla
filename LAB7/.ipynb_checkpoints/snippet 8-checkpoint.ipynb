time complexity: O(n²)
space complexity: O(1)

justify: the first loop takes n operations and the nested loops take n² operations so the total is dominated by n²

```python
def process(arr):
    n = len(arr)

    for i in range(n):
        print(arr[i])

    for j in range(n):
        for k in range(n):
            print(arr[j], arr[k])


arr = [1, 2, 3]
process(arr)
```
