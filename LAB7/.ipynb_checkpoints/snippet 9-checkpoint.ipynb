time complexity: O(n)
space complexity: O(1)

justify: the inner loop always runs only 10 times which is constant so the total work is 10n and becomes O(n)

```python
def check_first_ten(arr):
    for i in range(len(arr)):
        for j in range(10):
            if arr[i] == j:
                return True
    return False


arr = [15, 20, 7, 25]
print(check_first_ten(arr))
```
