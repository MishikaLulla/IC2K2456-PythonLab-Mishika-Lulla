```python
def has_duplicate(arr):
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] == arr[j]:
                return True

    return False


numbers = [10, 20, 30, 20, 40]

result = has_duplicate(numbers)

print("has duplicate:", result)
```

# snippet 2

time complexity: O(n²)
space complexity: O(1)

justify: there are two nested loops and in the worst case almost every pair of elements is compared
