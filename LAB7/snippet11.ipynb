```python
def reverse_in_place(arr):
    left, right = 0, len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1

    return arr


arr = [1, 2, 3, 4, 5]

result = reverse_in_place(arr)

print(result)
```

time complexity: O(n)
space complexity: O(1)

justify: the loop swaps elements from both ends and processes about n divided by 2 elements while using only a few variables
