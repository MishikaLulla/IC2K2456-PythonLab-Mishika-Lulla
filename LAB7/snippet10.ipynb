```python
def reverse_new(arr):
    reversed_arr = []

    for i in range(len(arr) - 1, -1, -1):
        reversed_arr.append(arr[i])

    return reversed_arr


arr = [1, 2, 3, 4, 5]

result = reverse_new(arr)

print(result)
```

time complexity: O(n)
space complexity: O(n)

justify: every element is copied once into a new list so both the running time and extra space grow with n
