```python
def count_pairs_with_sum(arr, target):
    seen = set()
    count = 0

    for num in arr:
        if target - num in seen:
            count += 1

        seen.add(num)

    return count


arr = [1, 5, 7, -1, 5]
target = 6

result = count_pairs_with_sum(arr, target)

print(result)
```

time complexity: O(n)
space complexity: O(n)

justify: the array is traversed once and the set can store up to n different elements
