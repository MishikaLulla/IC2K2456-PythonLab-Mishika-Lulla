```python
def find_max(arr):
    max_val = arr[0]

    for i in range(1, len(arr)):
        if arr[i] > max_val:
            max_val = arr[i]

    return max_val


numbers = [10, 25, 7, 40, 15]

result = find_max(numbers)

print("largest element:", result)
```

snippet 1

time complexity: O(n)
space complexity: O(1)

justify: the loop checks each element of the array once so the time increases linearly while only one extra variable is used
