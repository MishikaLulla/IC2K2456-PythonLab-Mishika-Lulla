```python
# QUESTION 3- this program checks whether a number is a perfect number
# and prints all perfect numbers up to a given limit
def is_perfect(number):
    if number <= 0:
        return False
    divisor_sum = 0
    for i in range(1, number):
        if number % i == 0:
            divisor_sum = divisor_sum + i
    return divisor_sum == number
number = int(input("Enter a number"))
if number <= 0:
    print(" enter a positive number")
elif is_perfect(number):
    print(number, "is a perfect number")
else:
    print(number, "is not a perfect number.")
limit = int(input("\nEnter the limit"))
if limit <= 0:
    print(" enter a positive number")
else:
    print("perfect numbers up to", limit, "are:")
    for i in range(1, limit + 1):
        if is_perfect(i):
            print(i)
```
