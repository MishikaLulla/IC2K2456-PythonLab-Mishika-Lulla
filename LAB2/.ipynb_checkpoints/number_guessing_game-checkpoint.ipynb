```python
# QUESTION 8 -this program creates a number guessing game with limited attempts

import random
secret_number = random.randint(1, 100)
maximum_attempts = 7
attempts = 0

print("welcome to the number guessing game!")
print("guess the number between 1 and 100")
print("you have", maximum_attempts, "attempts")


while attempts < maximum_attempts:

    guess = int(input("\nenter your guess "))

    if guess < 1 or guess > 100:
        print("please enter a number between 1 and 100")
        continue

    attempts += 1

    if guess < secret_number:
        print("too low!")

    elif guess > secret_number:
        print("too high!")

    else:
        print("correct!")
        print("you guessed the number in", attempts, "attempt(s)")
        break

else:
    print("\nyou have used all your attempts.")
    print("The correct number was:", secret_number)
```
