```python
# program 1. Variable and Identifier Practice Declare variables for your name, age, height and whether you are a student (True/False)
#This program demonstrates variables and their data types
```

```python
name = "Mishika"
age = 20
height = 5.4
is_student = True
print("Name:", name, "Type:", type(name))
print("Age:", age, "Type:", type(age))
print("Height:", height, "Type:", type(height))
print("Student:", is_student, "Type:", type(is_student))
```
