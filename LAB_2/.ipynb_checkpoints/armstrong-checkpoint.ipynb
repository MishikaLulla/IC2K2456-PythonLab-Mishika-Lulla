```python
# QUESTION 1 -this program checks whether a number is an armstrong number
# and prints all armstrong numbers within a given range

def is_armstrong(number):
    if number < 0:
        return False
    power = len(str(number))
    original_number = number
    total = 0
    while number > 0:
        digit = number % 10
        total = total + digit ** power
        number = number // 10
    if original_number == 0:
        total = 0

    return total == original_number
number = int(input("Enter a number: "))
if number < 0:
    print("Please enter a non-negative number.")
else:
    if is_armstrong(number):
        print(number, "is an Armstrong number.")
    else:
        print(number, "is not an Armstrong number.")

start = int(input("\nEnter the starting number of the range: "))
end = int(input("Enter the ending number of the range: "))
if start < 0 or end < 0:
    print("Please enter non-negative numbers for the range.")
elif start > end:
    print("Starting number cannot be greater than ending number.")
else:
    print("Armstrong numbers are:")

    for i in range(start, end + 1):
        if is_armstrong(i):
            print(i)
```
