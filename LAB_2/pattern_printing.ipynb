```python
# QUESTION 6 -this program prints different patterns using nested loops
def star_triangle(n):
    print("\nright angled triangle")
    for i in range(1, n + 1):
        for j in range(i):
            print("*", end="")
        print()
def number_pattern(n):
    print("\nnumber pattern:")
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            print(j, end="")
        print()
def pyramid(n):
    print("\npyramid pattern:")
    for i in range(1, n + 1):
        for j in range(n - i):
            print(" ", end="")
        for j in range(2 * i - 1):
            print("*", end="")
        print()
n = int(input("enter the number of rows "))
if n <= 0:
    print("please enter a positive number.")
else:
    star_triangle(n)
    number_pattern(n)
    pyramid(n)
```
